```python
import pandas as pd
df = pd.read_csv('people.csv')
```

```python
print(df)
```

SORTING


```python
print(df.sort_values('Age'))
```

Default sort_values(column_name) sorts the values in the ascending order

```python
print(df.sort_values('Age' , ascending = False))
```

ADDING A NEW COLUMN


```python
df['AgeinMonths'] = df['Age'] * 12
print(df)
```

New column named AgeinMonths created. New column values = Existing column values multiplied by 12. It's a math equation. 

RENAME a column


```python
df.rename(columns = {'City' : 'Location'}, inplace = True)
print(df)
```
